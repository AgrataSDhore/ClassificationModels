```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("customer_churn.csv")
```

```python
df
```

```python
df.info()
```

```python
# df['TotalCharges']=df['TotalCharges'].astype(float)  will throw an erroor
df['TotalCharges']=pd.to_numeric(df['TotalCharges'],errors='coerce')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
for col in df.columns:
  if df[col].dtype!="object":
    plt.boxplot(df[col])
    plt.title(col)
    plt.show()
```

```python
df.drop(['customerID'],axis=1,inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
for col in df.columns:
  if(df[col].dtype=="object"):
    df[col]=le.fit_transform(df[col])
```

```python
df
```

```python
df.info()
```

```python
# Model building
# 1.spliting the data
# 2.create model
# 3.train the model
# 4.predictions
# 5.evaluate the model
```

```python
x = df.iloc[:,:-1]
y = df['Churn']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=34)
```

```python
x_train
```

```python
y_train
```

```python
x_test
```

```python
y_test
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model=RandomForestClassifier(n_estimators=50,random_state=23)
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import *
```

```python
accuracy_score(y_test,y_pred)
```

```python
cm=confusion_matrix(y_test,y_pred)
```

```python
sns.heatmap(cm,cmap='Greens',annot =True,fmt="d")
```

```python
n=[50,100,150,200,250,300,350,400,450,500,550,600,650,700]
```

```python
for i in n:
  model=RandomForestClassifier(n_estimators=i,random_state=23)
  model.fit(x_train,y_train)
  y_pred=model.predict(x_test)
  acc=accuracy_score(y_test,y_pred)
  print(f"for the n_estmator {i} the accuracy_score is :{acc*100}")
```

```python
param_grid={
    "n_estimators":[100,200,300],
    "max_depth":[1,5,10],
    "min_samples_split":[2,5,7],
    "min_samples_leaf":[1,2,4],
    "criterion":["gini","entropy"]
}
```

```python
from sklearn.model_selection import GridSearchCV
model=RandomForestClassifier(random_state=23)

grid_search=GridSearchCV(estimator=model,param_grid=param_grid,cv=2)
```

```python
grid_search.fit(x_train,y_train)
```

```python
print(grid_search.best_params_)
```

```python
final_model=RandomForestClassifier(n_estimators=100,max_depth=10,min_samples_leaf=4,min_samples_split=2,criterion='entropy')
```

```python
final_model.fit(x_train,y_train)
```

```python
final_y_pred=final_model.predict(x_test)
```

```python
accuracy_score(y_test,final_y_pred)*100
```

```python
confusion_matrix(y_test,final_y_pred)
```
